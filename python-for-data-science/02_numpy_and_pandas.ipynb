## NumPy

NumPy është një **Python library** për punë me data numerike.
Struktura kryesore quhet **array** dhe është e optimizuar për llogaritje.

## NumPy Arrays

Një NumPy array:
- është koleksion vlerash numerike
- zakonisht me të njëjtin tip
- është i shpejtë për llogaritje

Shembull: temperatura ditore.

```python
temperaturat = np.array([20, 22, 19, 21, 23, 24, 18])
temperaturat
```

```python
print("Mesatarja:", temperaturat.mean())
print("Minimumi:", temperaturat.min())
print("Maksimumi:", temperaturat.max())
print("Numri i vlerave:", temperaturat.size)
```

## Pse jo vetëm listat?

Listat janë shumë të mira për shembuj të thjeshtë.
Por për analiza numerike me shumë të dhëna:

- NumPy arrays janë më të shpejtë
- janë të optimizuar për llogaritje
- punojnë mirë me shumë vlera njëkohësisht

(Në praktikë, shumë operacione në Pandas bazohen në NumPy.)

```python
# Shembull: mbledhja e listave
temp_list1 = [21, 23, 20, 22]
temp_list2 = [24, 25, 19, 20]

temp_list1+temp_list2
```

```python
# Shembull: mbledhja e array
temp_array1 = np.array(temp_list1)
temp_array2 = np.array(temp_list2)

temp_array1 + temp_array2
```

```python
# Shembull: rrisim çdo temperaturë me +1 gradë
temperaturat_plus1 = temperaturat + 1
temperaturat_plus1
```

```python
# Diferenca mes vlerave dhe mesatares
diferenca = temperaturat - temperaturat.mean()
diferenca
```

## DataFrame dhe Series

- **DataFrame**: tabelë me rreshta dhe kolona
- **Series**: një kolonë e vetme e një DataFrame

```python
data = {
    "Emri": ["Ana", "Jon", "Sara", "Elio", "Dora", "Ardit"],
    "Mosha": [25, 30, 22, 28, 35, 27],
    "Departamenti": ["IT", "Finance", "IT", "HR", "Finance", "IT"],
    "Paga": [800, 1200, 700, 1000, 1500, 950]
}

df = pd.DataFrame(data)
df
```

## Grupim: mesatarja sipas departamentit
Kjo është një nga operacionet më të përdorura në praktikë.


```python
data = {
    "Emri": ["Ana", "Jon", "Sara", "Elio", "Dora", "Ardit"],
    "Mosha": [25, 30, 22, 28, 35, 27],
    "Departamenti": ["IT", "Finance", "IT", "HR", "Finance", "IT"],
    "Paga": [800, 1200, 700, 1000, 1500, 950]
}

df = pd.DataFrame(data)
df.groupby(["Departamenti", 'Mosha'])["Paga"].mean()
```
