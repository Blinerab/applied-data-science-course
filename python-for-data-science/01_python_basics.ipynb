## Listat

Listat janë koleksione të renditura të elementeve.
Indeksimi fillon nga 0.

```python
filma = ["Inception", "Matrix", "Interstellar", "Fight Club", "Avatar"]

print(filma)
print("Filmi i parë:", filma[0])
print("Filmi i fundit:", filma[-1])
```

append(): Shton një element të vetëm në fund të listës.

```python

filma.append("Joker")
print("Pas append:", filma)
```

insert(): Vendos një element të vetëm në një pozicion të caktuar në listë

```python
filma.insert(0, "Titanic")
print("Pas insert:", filma)
```

remove(): Fshin një element të caktuar nga lista

```python
filma.remove("Matrix")
print("Pas remove:", filma)
```

pop(): Fshin dhe kthen elementin në një indeks të caktuar (parazgjedhja është elementi i fundit)

```python
filma.pop()
print("Pas pop(element i fundit by default):", filma)

# filma.pop(0)
# print("Pas pop(index=0):", filma)
```

extend(): Zgjeron listën duke shtuar të gjitha elementet nga një iterues tjetër (p.sh., një listë tjetër) në fund të listës

```python
filma.extend(filma)
print(filma)
```

Listat mund t'i modifikojmë gjithashtu duke përdorur veprimet matematikore

```python
filma1 = ["Inception", "Matrix", "Interstellar"]
filma2 = ["Fight Club", "Avatar"]

print(filma1 + filma2)
```

Për të marrë një pjesë të listës, duhet të përdorim metodën më poshtë

```python
filma = ["Inception", "Matrix", "Interstellar", "Fight Club", "Avatar"]
sliced_list = filma[1:4]
print(sliced_list)
```
